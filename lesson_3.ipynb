```python
def collect_scores():
  print("wo shi function ming collcet scores()")
  scores_array = []

  while True:
    user_input = input("qing shu ru cheng ji: ")

    if user_input == 'q':
      break

    score = int(user_input)
    scores_array.append(score)

  return scores_array

def print_scores(data_list):
  print("cheng ji qing dan === , wo shi function print_scores")
  for i in range(len(data_list)):
    print(f"di {i+1} wei xue sheng de cheng ji: {data_list[i]} fen")

def calculate_average(data_list):
  if len(data_list) == 0:
    print(" mei you cheng ji ke suan ping jun. ")
    return

  total = sum(data_list)
  count = len(data_list)
  average = total/count

  print("-" * 20)
  print(f"quan ban zong fen: {total} fen")
  print(f"quan ban ping jun: {average} fen")


math_scores = collect_scores()
print_scores(math_scores)

calculate_average(math_scores)
```
